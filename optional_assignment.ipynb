```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

```python

# Setting up the parameters for a normal distribution 
duration = 10
fps = 30
total_frames = fps * duration
samples_per_frame = 20
mean = 0
std = 1
```

```python
np.random.seed(42)
measurements = np.random.normal(mean, std, total_frames * samples_per_frame)
```

```python
# Creating the chart
fig, ax = plt.subplots(figsize=(10,6))
bins = np.linspace(-4, 4, 40)
curve = np.linspace(-4, 4, 200)

text = ax.text(0.03, 0.90, '', transform=ax.transAxes, fontsize=11, verticalalignment = 'top')

def update(frame):
    for patch in list(ax.patches):
        patch.remove()

    count = (frame+1) * samples_per_frame
    data = measurements[:count]

    # draw normalized histogram 
    ax.hist(data, bins=bins, density=True, color='blue', edgecolor='black', alpha=0.7)
    elapsed_time = (frame+1)/fps 
    ax.set_title(f"Building Normal Distribution ({elapsed_time: .1f}s/{duration}s)")
    text.set_text(f"Total Measurements: {count: ,}\nSample Mean: {np.mean(data):.3f}\nStandard Deviation: {np.std(data): .3f}")

#Generate animation
anim = FuncAnimation(fig, update, frames=total_frames, interval=1000/fps, blit=False)

plt.close(fig)
HTML(anim.to_jshtml())
```
